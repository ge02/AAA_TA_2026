## Raw Chicago Trip Data

First we sample the raw trip data downloaded from the webportal of the city of chicago: https://data.cityofchicago.org/Transportation/Taxi-Trips-2024-/ajtu-isnz/about_data (downloaded on 05.05.2026).

The data fetching can be found here: [Raw Chicago Trip Data](01_Preprocessing/01_01_Fetch_Trip_Data.ipynb)


```python
# # # # # # # # # # # # # # # # # # # # # #
#                                         #
# Load raw data                           #
#                                         #
# # # # # # # # # # # # # # # # # # # # # #

raw_trip_data = pd.read_csv("data/trips/CHICAGO_TAXI_TRIPS_2025.csv")
raw_trip_data.head()
```

```python
# # # # # # # # # # # # # # # # # # # # # #
#                                         #
# Sample raw trip data                    #
#                                         #
# # # # # # # # # # # # # # # # # # # # # #

SAMPLE_FRAC = 0.01

trip_sample = raw_trip_data.sample(frac=SAMPLE_FRAC, random_state=42)

trip_sample.to_csv("data/sample/SAMPLE_CHICAGO_TAXI_TRIPS_2025.csv", index=False)

print(f"Sampled {len(trip_sample):,} rows ({SAMPLE_FRAC:.3%} of total)")
print(f"Saved to data/sample/SAMPLE_CHICAGO_TAXI_TRIPS_2025.csv")
```
