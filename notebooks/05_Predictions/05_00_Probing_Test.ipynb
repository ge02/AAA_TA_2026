# 5.0 Representation-Probing Test

We have asked ourselves the question: since we should predict the demand at time t based on "generic" features like this: "what demand can we expect on a rainy Sunday from 10 – 11 am around the outskirts of the city" and not based on the demand at time t-1, can we use a random shuffle when splitting the data or should we keep a chronological split to avoid data leakage?.

To tackle this question we stuck to this definition of data leackage: "Data leakage is information about the target that leaks into your features/training data but wouldn't legitimately be available at prediction time" (Kaufman et al., 2012).

In our dataset, we have each time bucket (e.g. Monday 10am-11am) multiple times, once for each week. So in other words, if our Neural Networks (NNs) are able to implicitly learn to reconstruct the date or even the week of each e.g. Monday 10am-11am in our dataset, we need to split the data chronologically, if our NNs cannot identify the date or week, we can keep the train/test split with a random shuffle.

To test wether we can keep the random shuffle or not, we tested our NNs with probing.
