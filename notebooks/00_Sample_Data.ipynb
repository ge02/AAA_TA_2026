# 0 Sample Data Generation

In this notebook we generate sample data for each state of both the trip and weather (and merged) datasets for an easier and faster exemplary compliation of our notebooks. Please keep in mind that we sample the data randomly, therefore the sampled data is solely an exemplary (not perfect) representation of our datasets and some patterns etc. are not represented in the sampled datasets.

For the sampling we decided to sample the datasets down to a total of approx. 10.000 - 15.000 rows, keeping all columns.

All of our data can be found in the following sciebo folder: https://uni-koeln.sciebo.de/s/z5xgdGosd9y9c4e (Password: AAA2026)

```python
# # # # # # # # # # # # # # # # # # # # # #
#                                         #
# Import packages                         #
#                                         #
# # # # # # # # # # # # # # # # # # # # # #

import pandas as pd

# reset working dir
import os
from pathlib import Path
```

```python
# # # # # # # # # # # # # # # # # # # # # #
#                                         #
# Reset working directory                 #
#                                         #
# # # # # # # # # # # # # # # # # # # # # #

import __main__
_nb = getattr(__main__, "__vsc_ipynb_file__", None) or os.environ.get("JPY_SESSION_NAME")
_start = Path(_nb).resolve().parent if _nb else Path.cwd()
os.chdir(next(p for p in [_start, *_start.parents] if (p / "pyproject.toml").exists()))
print(f"Working directory: {os.getcwd()}")
```

```python
# # # # # # # # # # # # # # # # # # # # # #
#                                         #
# Configs                                 #
#                                         #
# # # # # # # # # # # # # # # # # # # # # #

SAMPLE_FRAC = 0.002
random_seed = 42
OUTPUT_DIR = "data/samples"
```

```python
# # # # # # # # # # # # # # # # # # # # # #
#                                         #
# Sampling function                       #
#                                         #
# # # # # # # # # # # # # # # # # # # # # #

def sample_merged_data(df, frac, random_state, output_file_name):
    output_path = f"{OUTPUT_DIR}/{output_file_name}"
    sample = df.sample(frac=frac, random_state=random_state)
    sample.to_parquet(output_path, index=False)
    print(f"Sampled {len(sample):,} rows ({frac:.2%} of total)")
    print(f"Saved to {output_path}")
    return sample
```
